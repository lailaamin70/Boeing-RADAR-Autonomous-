# TruckScenes Vehicle Velocity and Sensor Information Efficiency Investigation

**Dataset:** TruckScenes v1.2-mini


## 1. Project Goal

This investigation compares RADAR and LiDAR vehicle-motion sensing in TruckScenes under a common annotation-derived reference-velocity framework.

The project focuses on two related aspects:

1. vehicle velocity performance
2. sensor information efficiency

Velocity performance is evaluated in terms of velocity-estimation availability and accuracy relative to the common reference velocity. Sensor information efficiency is evaluated in terms of information availability, information amount, information type, velocity-estimation yield, and native observation frequency.

The current analysis is developed and validated using TruckScenes v1.2-mini. The processing workflow is structured so that the same analysis logic can later be rerun on the complete TruckScenes dataset.


## 2. Definitions

### 2.1 Vehicle Velocity

In this project, velocity refers to target-vehicle motion over time rather than individual sensor-point velocity.

Consecutive TruckScenes annotations are used to derive the common reference velocity. The 2D ground-plane speed is the primary vehicle reference quantity, while the corresponding 3D speed is retained mainly as a diagnostic measure.

RADAR and LiDAR provide vehicle-motion information through different measurement processes:

- RADAR returns provide direct relative-velocity components. After target association and ego-motion compensation, radial-velocity constraints are used to reconstruct a stable 2D target-velocity estimate in the global ground plane.
- LiDAR points do not contain a direct velocity field. Target position is estimated from target-associated LiDAR points in the global frame, and target velocity is derived from displacement over time between consecutive same-sensor observations.

The primary cross-sensor comparison therefore uses the validated 2D target-speed quantities from both modalities. A directly comparable validated RADAR 3D target-velocity estimate is not used in the main comparison.

### 2.2 Sensor Information Efficiency

In this project, efficiency refers to the amount and type of usable target-related sensor information acquired under equivalent observation conditions or time.

The efficiency analysis considers:

- target-information availability
- velocity-estimation availability
- velocity-estimation yield when target information is available
- target-associated information amount within each sensor modality
- information type
- native sensor observation frequency

RADAR returns and LiDAR points are different information units and are not treated as directly equivalent. Point or return counts are therefore analysed within each modality rather than combined into a single cross-sensor efficiency score.

This definition of efficiency does not refer to computational efficiency, file size, data loading speed, CPU/GPU usage, energy consumption, or hardware processing cost.


## 3. Research Questions

### 3.1 Main Research Question

How do RADAR and LiDAR compare in terms of vehicle velocity performance and usable sensor information under common annotation-derived vehicle-motion conditions in TruckScenes?

### 3.2 Velocity Questions

The velocity analysis considers the following questions:

- How can a common annotation-derived vehicle reference velocity be constructed?
- How can vehicle-level velocity be obtained from the validated RADAR and LiDAR measurements?
- What proportion of valid reference intervals produces a usable velocity estimate for each method?
- How closely do the available RADAR- and LiDAR-derived 2D speeds agree with the reference speed?
- How do the two methods compare on the same matched intervals where both velocity estimates are available?
- How do velocity availability and accuracy vary across reference-speed ranges?
- What methodological limitations affect the velocity comparison?

### 3.3 Efficiency Questions

The efficiency analysis considers the following questions:

- How often is target-associated RADAR or LiDAR information available under the common reference-interval conditions?
- How often does available target information produce a usable velocity estimate under the current sensor-specific methods?
- How does target-associated information amount vary across reference-speed ranges within each modality?
- What types of target-related information are provided by RADAR and LiDAR?
- How do native RADAR and LiDAR observation frequencies differ from the common keyframe-level comparison conditions?


## 4. Comparison Strategy

The primary analytical unit is one target vehicle over one valid consecutive annotation-derived reference interval.

All valid reference intervals are retained as the base comparison dataset. RADAR and LiDAR results are added using left joins so that intervals without a usable sensor-derived velocity estimate are not removed. Availability is treated as a research result rather than a filtering condition.

Velocity accuracy and velocity-estimation availability are analysed separately. A method is not considered more accurate simply because it produces estimates for more intervals.

Direct RADAR–LiDAR accuracy comparison is performed on the matched subset where both validated 2D velocity estimates are available under the same reference interval.

Vehicle-motion conditions are grouped using the common annotation-derived 2D reference-speed ranges:

- 0–<10 km/h
- 10–<20 km/h
- 20–<30 km/h
- 30–<40 km/h
- 40–<50 km/h
- 50–<60 km/h
- 60–<70 km/h
- 70–<80 km/h
- 80–<90 km/h
- 90–<100 km/h
- 100–<110 km/h
- ≥110 km/h

Sensor information efficiency is compared through multiple characteristics rather than a single score. Information availability, information amount, information type, and velocity-estimation yield are evaluated at the common reference/keyframe level. Native observation frequency is reported separately because the sensor-specific target-velocity analyses do not use all intermediate native observations.


## 5. Sensor Characteristics Relevant to the Investigation

The validated dataset inspection and sensor-processing stages establish several characteristics that determine how the comparison must be performed:

- RADAR PCD data provide 3D return positions and direct relative-velocity components (`vrel_x`, `vrel_y`, and `vrel_z`).
- LiDAR PCD data provide 3D point positions, intensity, and point-level timestamps, but no direct velocity field.
- Ego-motion metadata provides ego-vehicle velocity information required for RADAR ego-motion compensation.
- Annotation trajectories provide the common target-position history used to construct the reference velocity.
- RADAR target velocity and LiDAR target velocity require different validated processing methods before they can be compared at the vehicle level.
- Target-information availability is not equivalent to velocity-estimation availability for either modality.
- RADAR and LiDAR operate at different native observation frequencies, while the main target-level comparison is performed using common annotation/keyframe intervals.
- RADAR return counts and LiDAR point counts cannot be interpreted as equivalent information quantities.

These differences are treated explicitly in the later analysis rather than being reduced to a single sensor-performance metric.


## 6. Analysis Workflow

| Notebook | Purpose |
|---|---|
| `00_issue_overview.ipynb` | Define the project problem, comparison scope, research questions, and analysis strategy |
| `01_dataset_inspection.ipynb` | Inspect and validate the TruckScenes dataset and RADAR/LiDAR observation characteristics required by the later analysis |
| `02_annotation_reference_velocity.ipynb` | Generate and validate the common annotation-derived reference vehicle velocity |
| `03_radar.ipynb` | Process target-associated RADAR information and develop the validated RADAR-derived 2D target-velocity baseline |
| `04_lidar.ipynb` | Process target-associated LiDAR information and develop the validated LiDAR-derived target-velocity baseline |
| `05_comparison.ipynb` | Build the common interval-level dataset and compare RADAR and LiDAR velocity performance and sensor information efficiency |
| `06_final_summary.ipynb` | Summarise the complete workflow, key findings, limitations, and the procedure for extending the analysis to the full TruckScenes dataset |

Reusable processing logic validated in the analysis notebooks is stored in the Python modules under `src/`. Validated intermediate results are stored under `results/validated/` so that later analysis stages can reuse completed processing outputs without repeating earlier sensor-processing steps.
